# **Python คืออะไร? (What is Python?)**

Python คือภาษาโปรแกรมมิ่งระดับสูงที่ถูกนิยามด้วยคุณสมบัติหลัก 3 ประการ:

* High-level Language (ภาษาระดับสูง): ไวยากรณ์ (Syntax)
มีความใกล้เคียงกับภาษาอังกฤษ มนุษย์อ่านแล้วเข้าใจได้ทันที โดยที่คุณไม่ต้องไปวุ่นวายกับการจัดการหน่วยความจำ (Memory Management) ในระดับฮาร์ดแวร์เหมือนภาษาซี

* Interpreted Language (ภาษาประเภทอินเทอร์พรีเตอร์): คอมพิวเตอร์จะอ่านและทำงานตามโค้ดทีละบรรทัดจากบนลงล่าง ไม่ต้องเสียเวลารอ "คอมไพล์" (Compile) โค้ดทั้งหมดให้เป็นไฟล์ .exe ก่อนรัน ทำให้เขียนเสร็จแล้วกดสั่งรันดูผลลัพธ์ได้ทันที เหมาะมากกับงานต้องการความรวดเร็วอย่าง Scripting

* General-purpose Language (ภาษาเอนกประสงค์): ไม่ได้ถูกสร้างมาเพื่อสร้างเว็บหรือคำนวณคณิตศาสตร์เพียงอย่างเดียว แต่มันถูกออกแบบมาให้ทำได้ทุกอย่าง ขึ้นอยู่กับว่าคุณจะเอาไปใช้ทำอะไร

# Sample Puzzler Snippets:


```python
# Puzzler 1
x = 10
y = 3.5
z = x + y
print(z)
print(type(z))

# z จะเท่ากับ 13.5
# type ของ z ก็คือ float
# เนื่องจากว่า x = 10 y = 3.5  x มี type เป็น int  y มี type เป็น float
# เมื่อบวกกันคำตอบที่ได้จึงเป็น type float
```

```python
# Puzzler 2
name = "Alice"
age = "30" # Note: This is a string
message = name + " is " + age + " years old."
print(message)
print(type(age))

# คำตอบคือ Alice is 30 years old.
# type ของ age ก็คือ int
```

```python
# Puzzler 3
is_sunny = True
is_raining = False
print(is_sunny and is_raining)
print(type(is_sunny))

# คำตอบคือ False
# type ของ is_sunny ตือ boolean
```

```python
# Puzzler 4
# What happens here?
num_str = input("Enter a number: ")
result = num_str + 5
print(result)

# ตัวเลขที่ใส่เข้าไป จะเพิ่มขึ้น 5
# oof error เนื่องจากว่า ค่าที่ใส่เข้าไปเป็น str จึงบวกกัน int ไม่ได้
# สรุปข้อนนี้คือ error เนื่องจาก str จึงบวกกัน int ไม่ได้
```

# Variables and Basic Data Types

```python
print("=" * 50)
print(" "*((50 - len("PYTHON DATA TYPES EXAMPLE")) // 2)+ "PYTHON DATA TYPES EXAMPLE")
print("=" * 50)

# =========================
# Integers
# =========================
age = 30
print("\n[INTEGER]")
print("My age is:", age)
print("Type of age:", type(age))

# =========================
# Floats
# =========================
price = 19.99
print("\n[FLOAT]")
print("The price is:", price)
print("Type of price:", type(price))

# =========================
# Strings
# =========================
name = "Alice"
greeting = "Hello"
message = "Python is fun!"

print("\n[STRING]")
print(greeting + ", " + name + "!")
print(message)
print("Type of name:", type(name))

# Multi-line string
long_text = """
This is a multi-line string.
It can span across several lines.
Useful for longer descriptions.
"""
print(long_text)

# =========================
# Booleans
# =========================
is_student = True
is_adult = False

print("[BOOLEAN]")
print("Am I a student?", is_student)
print("Am I an adult?", is_adult)
print("Type of is_student:", type(is_student))

print("\n" + "=" * 50)
print(" "*((50 - len("END OF PROGRAM")) // 2)+ "END OF PROGRAM")
print("=" * 50)
```

# Input and Output

```python
# ==========================================
#           INPUT & OUTPUT DEMO
# ==========================================

print("╔══════════════════════════════╗")
print("║      INPUT / OUTPUT DEMO     ║")
print("╚══════════════════════════════╝")

# ==========================================
# Basic Output
# ==========================================
print("\n[OUTPUT EXAMPLES]")
print("This is a simple message.")
print("Python is fun!", end=" ")
print("Let's learn together.")
print("Item 1", "Item 2", "Item 3", sep=" | ")

# ==========================================
# User Input
# ==========================================
print("\n[USER INFORMATION]")

name = input("👤 Please enter your name: ")
print(f"Hello, {name}!")

# ==========================================
# Age Calculator
# ==========================================
age_str = input("🎂 What is your age? ")

print(f"You entered: {age_str}")
print(f"Data type: {type(age_str)}")

age_int = int(age_str)
next_year_age = age_int + 1

print(f"Next year, you will be {next_year_age} years old.")

# ==========================================
# Price Calculator
# ==========================================
price_str = input("💲 Enter a price (e.g., 19.99): ")

price_float = float(price_str)
tax = price_float * 0.07
total_price = price_float + tax

print("\n========== RECEIPT ==========")
print(f"Original price : ${price_float:.2f}")
print(f"Tax (7%)       : ${tax:.2f}")
print(f"Total price    : ${total_price:.2f}")
print("=============================")

print("\n Program Finished.")
```
