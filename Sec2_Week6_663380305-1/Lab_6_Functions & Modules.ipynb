### Activity 6.1: Refactoring a Previous Lab Assignment

```python
# refactored_classifier.py

def is_positive_negative_zero(num: float) -> str:
    """Classifies a number as positive, negative, or zero."""
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    return "Zero"

def is_even_odd(num: int) -> str:
    """Determines whether an integer is even or odd."""
    if num % 2 == 0:
        return "Even"
    return "Odd"

def process_number(num: float):
    """Encapsulates the full classification and prints the summary."""
    sign = is_positive_negative_zero(num)
    
    # Parity check only applies to integers
    if num.is_integer():
        parity = is_even_odd(int(num))
        print(f"Result for {int(num)}: {sign}, {parity}")
    else:
        print(f"Result for {num}: {sign}, Non-integer (N/A)")

def main():
    """Main execution loop for user interaction."""
    print("--- Refactored Number Classifier ---")
    while True:
        user_input = input("\nEnter a number (or 'q' to quit): ").strip()
        if user_input.lower() == 'q':
            print("Exiting classifier. Goodbye!")
            break
        
        try:
            val = float(user_input)
            process_number(val)
        except ValueError:
            print("Invalid input! Please enter a valid numerical value.")

if __name__ == "__main__":
    main()
```

### Activity 6.2: Modular Programming & Standard Library Exploration

#### Benefits of Modular Programming

| Benefit | Practical Impact |
| :--- | :--- |
| **Maintainability & Debugging** | Bugs are isolated to specific functions or modules without breaking the rest of the application. |
| **Code Reusability (DRY)** | Encapsulating logic in functions prevents redundant, repeated code. |
| **Teamwork & Collaboration** | Team members can work concurrently on separate modules in larger projects. |
| **Testability** | Functions can be unit-tested individually with mock inputs. |

#### Standard Library Exploration

Exploration of 3 standard library modules from the official Python documentation (`docs.python.org/3/library/`):

* **`datetime`**: Used for parsing, formatting, and manipulating timestamps.
* **`os`**: Used for interacting with operating system paths and directories.
* **`collections`**: Provides specialized datatypes like `Counter` for sequence counting.

```python
# Standard Library Module 1: datetime
from datetime import datetime

current_time = datetime.now()
print("Formated Timestamp:", current_time.strftime("%Y-%m-%d %H:%M:%S"))

# Standard Library Module 2: os
import os

print("Working Directory:", os.getcwd())

# Standard Library Module 3: collections
from collections import Counter

data_samples = ["apple", "banana", "apple", "cherry", "banana", "apple"]
counter_result = Counter(data_samples)
print("Most Frequent Item:", counter_result.most_common(1))
```
