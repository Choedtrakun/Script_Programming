***Format***: Individual work with peer discussion/instructor led review.

***Instructions***: Provide several code snippets with if/elif/else and logical operators. Students must trace the execution step-by-step for given input values and determine the final output.

```python
# Snippet A
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"

print(f"Grade: {grade}")

# Output is "B"
```

```python
# Snippet B
temp = 25
is_raining = True

if temp > 20 and not is_raining:
    print("Perfect day for a picnic!")
elif temp <= 20 or is_raining:
    print("Stay indoors or bring an umbrella.")
else:
    print("Unsure about the weather.")

# Output is "Stay indoors or bring an umbrella."
```

```python
# Snippet C
age = 17
has_license = True

if age >= 18:
    if has_license:
        print("You can drive legally.")
    else:
        print("You need a license to drive.")
else:
    print("You are too young to drive.")

# Output is "You are too young to drive."
```

```python
# Block B: Vectorized Filter

phrase = "Python 2026 Engine"

consonants = {
    c.lower()
    for c in phrase
    if c.isalpha() and c.lower() not in 'aeiou'
}

print(f"Unique consonants: {consonants}")

"""
เป็น If-Else ใน for loop ที่ If-Else มี 3 เงื่อนใข c.isalpha() และ c.lower() และต้องไม่อยู่ใน(หรือไม่เป็นสะระใน ENG) aeiou
3 เงื่อนใขเป็นจริงจึงจะทำงาน

loop ครั้งที่ 1 จะได้ h
loop ครั้งที่ 2 จะได้ g
loop ครั้งที่ 3 จะได้ n
loop ครั้งที่ 4 จะได้ t
loop ครั้งที่ 5 จะได้ p
loop ครั้งที่ 6 จะได้ y
และจบการทำงาน
"""
```

## **Activity 2.2**: Design a Simple "Choose Your Own Adventure" Text Game

```python
# Idea
"""
เกมทายตัวเลขโดยตัวเกมจะสุ่มเลขให้เราจาก 1 - 100
แล้วให้เราทายไปเรื่อยๆจนกว่าจะถูกต้อง
ซึ่งผู้เล่นจะทายได้แค่ 5 ครั้ง

โดยสมมุต
เกมสุ่มได้เลข 53
เราทาย 56 >> เกมจะบอกเราว่า 56 เป็นเลขที่สูงไป
เราทายอีกครั้ง 47 >> เกมจะบอกว่า 57 เป็นเลขที่ต่ำไป

หากตัวเลขูกต้อง ก็ชนะ หากยังผิด 5 ครั้งก็แพ้
"""
```

```python
# Raw Coding
import random
numrand = int((random.random())*100)

heart = 5
guess = 1
print("====Guess Number Game====")

while heart != 0:
    print(f"=== Guess {guess} ===")
    user_input = input("Enter your guess: ")
    try:
        user_input = int(user_input)
        if user_input == numrand:
            print("You are WIN!!!!!")
            print(f"The number is {numrand}")
            break
        elif user_input > numrand:
            print(f"{user_input} is higher than the number")
            heart -= 1
            print(f"You have {heart} heart(s) left")
        elif user_input < numrand:
            print(f"{user_input} is lower than the number")
            heart -= 1
            print(f"You have {heart} heart(s) left")
        elif heart == 0:
            print("You are LOSE!!!!!")
            print(f"The number is {numrand}")
        guess += 1
    except ValueError:
        print("Your guess must be the number")

    print()
```

```python
# Coding more beautiiful by ChatGPT
import random
numrand = int((random.random())*100)

heart = 5
guess = 1

print("=" * 35)
print("🎯      GUESS NUMBER GAME      🎯")
print("=" * 35)
print("Guess the number between 1 - 100")
print(f"❤️ Hearts: {heart}")
print()

while heart > 0:
    print("-" * 35)
    print(f"🎲 Guess #{guess}")
    print("-" * 35)

    user_input = input("👉 Enter your guess: ")

    try:
        user_input = int(user_input)

        if user_input == numrand:
            print("\n🎉 CONGRATULATIONS! YOU WIN! 🎉")
            print(f"🏆 The number was: {numrand}")
            break

        elif user_input > numrand:
            heart -= 1
            print(f"📈 {user_input} is TOO HIGH!")

        else:
            heart -= 1
            print(f"📉 {user_input} is TOO LOW!")

        if heart > 0:
            print(f"❤️ Hearts left : {heart}")
            print("🔄 Try again!\n")
        else:
            print("\n💀 GAME OVER 💀")
            print(f"🎯 The correct number was: {numrand}")

        guess += 1

    except ValueError:
        print("❌ Please enter a valid integer!\n")
```
