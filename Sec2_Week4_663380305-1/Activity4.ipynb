# **Activity 4.1:** List vs. Tuple: Scenario Matching
***Format***: Small group discussion, followed by a class debrief.

***Instructions***: Present students with a list of real-world data scenarios. For each
scenario, they must decide whether a list or a tuple would be the more appropriate
Python data structure and explain their reasoning (focus on mutability/immutability and
ordered nature).

***Scenarios***: 
1. collection of daily temperatures recorded over a week.
2. The RGB color values for a specific shade (e.g., (255, 0, 0) for red).
3. Items in a shopping cart that can be added or removed.
4. The days of the week in order.
5. Student scores on a series of assignments, which might be updated.
6. GPS coordinates for a fixed landmark.


***Purpose***: Reinforce understanding of when to use which data structure based on their
characteristics.


1. ข้อมูลอุณหภูมิประจำวันที่มีการบันทึกไว้ตลอดหนึ่งสัปดาห์ 
    - tuple 
    - ข้อมูลอุณหภูมิประจำวัน หากมีการแก้ใขก็จะไม่ใช่ของวันนั้น จึงไม่ควรมีการแก้ใข
2. ค่าสี RGB ของเฉดสีเฉพาะเฉดหนึ่ง (เช่น สีแดงคือ (255, 0, 0))
    - tuple
    - เป็นเพียงค่าสี RGB ของเฉดสีเฉพาะเฉดหนึ่ง ถ้าเปลี่ยนก็จะเป็ยเฉดอื่น จึงไม่จำเป็นต้องแก้ไข
3. รายการสินค้าในรถเข็นชอปปิงที่สามารถกดเพิ่มหรือลบออกได้ 
    - list
    - รายการสินค้าในรถเข็นชอปปิง ต้องรองรับการเพิ่มหรือลบออกได้
4. รายชื่อวันในหนึ่งสัปดาห์เรียงตามลำดับ (จันทร์ - อาทิตย์)
    - tuple
    - รายชื่อวันในหนึ่งสัปดาห์เป็นวันเดิมเสมอไม่มีการเปลียนแปลง จึงไม่จำเป็นต้องแก้ไข
5. คะแนนเก็บของนักเรียนจากการส่งงานชุดต่างๆ ซึ่งอาจมีการอัปเดตหรือแก้ไขคะแนนในภายหลังได้
    - list
    - ต้องมีการอัปเดตหรือแก้ไขคะแนนภายหลัง
6. พิกัด GPS ของสถานที่สำคัญหรือจุดแลนด์มาร์กที่ตั้งอยู่กับที่
    - tuple
    - พิกัด GPS ของสถานที่สำคัญตั้งอยู่กับที่ จึงไม่จำเป็นต้องแก้ไข

***Instructions***: Students will implement a very basic text-based to-do list manager using
what they've learned about lists.

***Features to include (minimum)***: 
1. An empty list todo_list initially.
2. A while loop that presents a menu:
    - Add a task
    - View tasks

    - Exit
3. Use input() to get user choices.
4. Use list methods (append(), remove() or pop(), len()) to manage the todo_list.

***Purpose***: Practical application of list operations in a mini-project context, integrating
loops and conditionals.
